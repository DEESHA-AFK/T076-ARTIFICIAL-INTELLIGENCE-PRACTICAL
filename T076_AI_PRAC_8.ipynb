```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Transaction dataset
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Butter']
]

# Convert transactions into a DataFrame
te = TransactionEncoder()
data = te.fit(transactions).transform(transactions)

df = pd.DataFrame(data, columns=te.columns_)

# Find frequent itemsets
frequent_items = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent_items)

# Generate association rules
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.6
)

print("\nAssociation Rules:")
print(rules[['antecedents', 'consequents',
             'support', 'confidence']])

import matplotlib.pyplot as plt

rules['rule'] = rules['antecedents'].astype(str) + " -> " + rules['consequents'].astype(str)

plt.bar(rules['rule'], rules['confidence'])

plt.xlabel("Association Rules")
plt.ylabel("Confidence")
plt.title("Association Rules Confidence")

plt.xticks(rotation=45)

plt.show()
```
